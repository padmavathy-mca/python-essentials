```python
#Ecommerce Sales project
import numpy as np
import pandas as pd

#Create sales dataset
data = {"Product": ["Laptop","Phone","Headphones","Keyboard","Mouse","Monitor","Laptop","Phone","Headphones","Monitor","Keyboard","Mouse"],
        "Category": ["Electronics","Electronics","Accessories","Accessories","Accessories","Electronics","Electronics","Electronics","Accessories",
                     "Electronics","Accessories","Accessories"],
        "Quantity": [2, 5, 10, 8, 15, 3,1, 4, 12, 2, 6, 20],
        "Price": [55000, 30000, 2000, 1500,800, 25000, 55000, 30000,2000, 25000, 1500, 800],
        "Rating": [4.5, 4.2, 4.7, 4.1,4.6, 4.3, 4.8, 4.0,4.5, 4.4, 4.2, 4.7]
       }
df = pd.DataFrame(data)

print("\nOriginal Sales Data:")
print(df)
```

```python
#Understanding the data
print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```

```python
#Calculate Revenue by using numpy

df["Revenue"] = np.multiply(df["Quantity"],df["Price"])

print(df[["Product","Quantity","Price","Revenue"]])
```

```python
#Basic Numpy statistics
revenue = df["Revenue"].to_numpy()

print("Total Revenue:",np.sum(revenue))
print("Average Revenue:",np.mean(revenue))
print("Highest Revenue:",np.max(revenue))
print("Lowest Revenue:",np.min(revenue))
print("Revenue Standard Deviation:",np.std(revenue))
```

```python
#Find the highest revenue transaction
highest_index = np.argmax(revenue)
print(df.iloc[highest_index])
```

```python
#Product wise sales
product_sales = df.groupby("Product")["Quantity"].sum()
print(product_sales)
```

```python
# Find best-selling product
best_selling_product = product_sales.idxmax()
best_selling_quantity = product_sales.max()
print("\nBest Selling Product:",best_selling_product)
print("Units Sold:",best_selling_quantity)
```

```python
#Product wise revenue
product_revenue = df.groupby("Product")["Revenue"].sum()
print(product_revenue)

highest_revenue_product = product_revenue.idxmax()
highest_revenue_value = product_revenue.max()
print("\nHighest Revenue Product:",highest_revenue_product)
print("Revenue:",highest_revenue_value)
```

```python
#Category wise revenue
category_revenue = df.groupby("Category")["Revenue"].sum()
print(category_revenue)
```

```python
#Highest rating products
high_rated_products = df[df["Rating"] >= 4.5]
print(high_rated_products[["Product","Rating"]])
```

```python
average_rating = np.mean(df["Rating"].to_numpy())
print("\nAverage Product Rating:",round(average_rating, 2))
```

```python
#Find usually high revenue
mean_revenue = np.mean(revenue)
std_revenue = np.std(revenue)
threshold = mean_revenue + std_revenue
unusual_sales = df[df["Revenue"] > threshold]

print(unusual_sales[["Product","Quantity","Revenue"]])
```

```python
# Rating contributes up to 100 points
# Quantity contributes additional points
# Creates persormance score
df["Performance_Score"] = (df["Rating"] * 20 + df["Quantity"] * 2)
print(df[["Product","Rating","Quantity","Performance_Score"]])
```

```python
#Sort products by performance
ranked_products = df.sort_values("Performance_Score",ascending=False)
print(ranked_products[["Product","Performance_Score"]].head(5))
```

```python
#Numpy slicing
quantity_array = df["Quantity"].to_numpy()
print("All quantities:")
print(quantity_array)

print("\nFirst 5 quantities:")
print(quantity_array[:5])

print("\nLast 5 quantities:")
print(quantity_array[-5:])
```

```python
#Numpy Sorting
sorted_revenue = np.sort(revenue)
print("Revenue from lowest to highest:")
print(sorted_revenue)
```

```python
#Normalization demo
normalized_revenue = (revenue / np.max(revenue))
print(normalized_revenue)
```

```python
#Final business report
print("FINAL BUSINESS REPORT")
print("\nTotal Revenue: ₹",np.sum(revenue))
print("Average Transaction Revenue: ₹",round(np.mean(revenue), 2))
print("Best Selling Product:",best_selling_product)
print("Highest Revenue Product:",highest_revenue_product)
print("Highest Revenue Transaction: ₹",np.max(revenue))
print("Average Rating:",round(average_rating, 2))
print("Number of Highly Rated Transactions:",len(high_rated_products))
print("Unusually High Revenue Transactions:",len(unusual_sales))
```
