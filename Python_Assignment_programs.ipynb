**1. Write a program to remove lowercase substrings from a given string**

```python
s = input('Enter a string:')
result = ""
for ch in s:
    if not ch.islower():
        result += ch
print('substrings after removing lowercase characters are: ',result) 
```

**3. Write a program to insert spaces between words starting with capital letters**

```python
text = input('Enter a string: ')
result = text[0]
for char in text[1:]:
    if char.isupper():
        result += ' '
    result += char

print('Result: ',result)
```

**4. Program to remove the parenthesis area in a string**

```python
text = input('Enter a string: ')
result = ''
inside = False
for char in text:
    if char == '(':
        inside = True
    elif char == ')':
        inside = False
    elif not inside:
        result += char

print('Result String:',result)
```

**5. Write a program to split a string with multiple delimiters.**

```python
delimiters_list = [',',';','|',':']
text = 'python,ML|RAG;LLM'

for delimiter in delimiters_list:
    text = text.replace(delimiter,' ')
result = text.split()

print('Result ', result)
```

**6. Write a program to find all adverbs and their positions in a given sentence.**

```python
import spacy
nlp = spacy.load('en_core_web_sm')

def find_adverbs(sentence):
    doc = nlp(sentence)
    print(f"\nSentence: '{sentence}'")

    header_format = "{:<15} | {:<8}"
    print(header_format.format('Adverb','Position'))
    print('-'*25)
    
    found = False
    for token in doc:
        if token.pos_ == 'ADV' or token.dep_ == "npadvmod":
            print(header_format.format(token.text,token.idx))
            found = True

    if not found:
        print('No adverbs found')

sentence_1 = "Yesterday, he completely forgot his keys and always does."
sentence_2 = "The red car drove up the steep hill."

find_adverbs(sentence_1)
find_adverbs(sentence_2)
```

**7. Write a program to do a case-insensitive string replacement.**

```python
def case_insensitive_replace_text(text,old,new):
    result = []
    i=0
    text_lower = text.lower()
    old_lower = old.lower()
    old_len = len(old)

    while i < len(text):
        if text_lower[i: i+old_len] == old_lower:
            result.append(new)
            i += old_len
        else:
            result.append(text[i])
            i += 1
    return ''.join(result)

text = "Java is great. Java code and Java! I love Java."
old = "Java"
new = "Python"

result = case_insensitive_replace_text(text, old, new)
print("Original:", text)
print("Replaced:", result)
```

**8. Write a program to split a string at uppercase letters.**

```python
def split_at_uppercase(text):
    if not text:
        return []
    res = [text[0]]
    for char in text[1:]:
        if char.isupper():
            res.extend([' ',char])
        else:
            res.append(char)
    return ''.join(res).split(' ')

text = 'PythonProgrammingLanguage'
print('Split List:',split_at_uppercase(text))
```
