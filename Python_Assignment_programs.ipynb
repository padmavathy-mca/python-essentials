**1. Write a program to remove lowercase substrings from a given string**

```python
s = input('Enter a string:')
result = ""
for ch in s:
    if not ch.islower():
        result += ch
print('substrings after removing lowercase characters are: ',result) 
```

**2. Write a program that reads a given expression and evaluates it**

```python
expression = input('Enter an expression / Mathematical calculation: ')
result = eval(expression)
print('Result: ',result)
```

**3. Write a program to insert spaces between words starting with capital letters**

```python
text = input('Enter a string: ')
result = text[0]
for char in text[1:]:
    if char.isupper():
        result += ' '
    result += char

print('Result: ',result)
```

**4. Program to remove the parenthesis area in a string**

```python
text = input('Enter a string: ')
result = ''
inside = False
for char in text:
    if char == '(':
        inside = True
    elif char == ')':
        inside = False
    elif not inside:
        result += char

print('Result String:',result)
```
