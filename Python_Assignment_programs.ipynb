**4. Program to remove the parenthesis area in a string**

```python
text = input('Enter a string: ')
result = ''
inside = False
for char in text:
    if char == '(':
        inside = True
    elif char == ')':
        inside = False
    elif not inside:
        result += char

print('Result String:',result)
```

**6. Write a program to find all adverbs and their positions in a given sentence.**

```python
#pip install spacy
#!python -m spacy download en_core_web_sm
```

```python
import spacy
nlp = spacy.load('en_core_web_sm')

def find_adverbs(sentence):
    doc = nlp(sentence)
    print(f"\nSentence: '{sentence}'")

    header_format = "{:<15} | {:<8}"
    print(header_format.format('Adverb','Position'))
    print('-'*25)
    
    found = False
    for token in doc:
        if token.pos_ == 'ADV' or token.dep_ == "npadvmod":
            print(header_format.format(token.text,token.idx))
            found = True

    if not found:
        print('No adverbs found')

sentence_1 = "Yesterday, he completely forgot his keys and always does."
sentence_2 = "The red car drove up the steep hill."

find_adverbs(sentence_1)
find_adverbs(sentence_2)
```

**7. Write a program to do a case-insensitive string replacement.**

```python
def case_insensitive_replace_text(text,old,new):
    result = []
    i=0
    text_lower = text.lower()
    old_lower = old.lower()
    old_len = len(old)

    while i < len(text):
        if text_lower[i: i+old_len] == old_lower:
            result.append(new)
            i += old_len
        else:
            result.append(text[i])
            i += 1
    return ''.join(result)

text = "Java is great. Java code and Java! I love Java."
old = "java"
new = "Python"

result = case_insensitive_replace_text(text, old, new)
print("Original:", text)
print("Replaced:", result)
```

**8. Write a program to split a string at uppercase letters.**

```python
def split_at_uppercase(text):
    if not text:
        return []
    res = [text[0]]
    for char in text[1:]:
        if char.isupper():
            res.extend([' ',char])
        else:
            res.append(char)
    return ''.join(res).split(' ')

text = 'PythonProgrammingLanguage'
print('Split List:',split_at_uppercase(text))
```

**9. Write a program to remove everything except alphanumeric characters from a string.**

```python
def only_alphanumeric(text):
    alnum_chars = [char for char in text if char.isalnum()]
    return ''.join(alnum_chars)

input_str = 'Hello@Python_Programming 3.13!!!'
result = only_alphanumeric(input_str)

print("Original:", input_str)
print("Cleaned:", result)
```

**12. Write a program to convert snake case string to camel case string.**

```python
def snake_to_camel(text):
    words = text.split('_')
    camel_words = [words[0]]+[word.capitalize() for word in words[1:]]
    return ''.join(camel_words)

input_str = 'python_programming_language'
result = snake_to_camel(input_str)

print('Original Text: ',input_str)
print('CamelCase: ',result)
```

**13. Write a program to determine whether a given year is a leap year.**

```python
def is_leap_year(year):
    #True if divisible by 400, OR (divisible by 4 and NOT divisible by 100)
    return (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0)

#Test the program
years = [2000,2024,1900,2026]
for y in years:
    if is_leap_year(y):
        print(f"{y} is a leap year.")
    else:
        print(f"{y} is NOT a leap year.")
```

**14. Write a program to convert a string to datetime.**

```python
from datetime import datetime

def convert_str_to_date(s,f):
    
    try:
        # strptime parses the string based on the provided format
        return datetime.strptime(s, f)        
    except ValueError as e:
        print(f"Error: The string does not match the format provided. Details: {e}")
        return None


print("Basic YYYY-MM-DD")
date_str = "2026-08-16"
date_format = "%Y-%m-%d"
result = convert_str_to_date(date_str, date_format)
if result:
    print(f'Converted: {result}')
    print(f'Date only: {result.date()}')
    print(f'Year=>{result.year}, Month=>{result.month}, Day=>{result.day}\n')

print("ISO format (used by APIs)")
date_str = "2026-08-16T14:30:45"
date_format = "%Y-%m-%dT%H:%M:%S"
result = convert_str_to_date(date_str, date_format)
if result:
    print(f'Converted: {result}')
    print(f'Date only: {result.date()}')
    print(f'Year=>{result.year}, Month=>{result.month}, Day=>{result.day}')
```

**15. Write a function that accepts a string and calculate the number of upper case letters and lower case letters.**

```python
def count_case_letters(text):
    upper_count = 0
    lower_count = 0

    for char in text:
        if char.isupper():
            upper_count += 1
        elif char.islower():
            lower_count += 1

    print(f'Original String: {text}')
    print(f'Number of Uppercase characters: {upper_count}')
    print(f'Number of Lowercase characters: {lower_count}')

input_str = 'Hello World!'
count_case_letters(input_str)
```

**16. Write a function that takes a list and returns a new list with unique elements of the first list.**

```python
def get_unique_elements(input_list):
    unique_list = []
    for item in input_list:
        if item not in unique_list:
            unique_list.append(item)
    return unique_list
    return list(set(input_list))
original_list = [1,2,3,4,2,3,4,1,3,2,1,4]
unique_list = get_unique_elements(original_list)
print(f'Original list: {original_list}')
print(f'Unique list: {unique_list}')
```

**17. Write a function that takes a number as a parameter and check the number is prime or not.**

```python
def is_prime(number):
    # Numbers less than or equal to 1 are not prime
    if number <= 1:
        return False
        
    # Check for factors from 2 up to the square root of the number
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            return False  # Found a factor, so it is not prime
            
    return True  # No factors found, so it is prime


num = int(input('Enter a number to check whether it is prime or not: '))    
if is_prime(num):
    print(f'{num} is a prime number')
else:
    print(f'{num} is NOT prime number')
```

**18. Write a program to print the even numbers from a given list.**

```python
def even_num(num_list):
    return [num for num in num_list if num % 2 == 0]

sample_list = [11,23,54,67,65,32,99]
print(f'Even numbers are: {even_num(sample_list)}')
```

**19. Write a function to check whether a number is perfect or not.**

```python
def is_perfect_num(number):
    if number <= 0:
        return False
    div_sum = 0
    for i in range(1,number):
        if number % i == 0:
            div_sum += i

    return div_sum == number

n = int(input('Enter a number to check whether it is perfect number or not: '))
if is_perfect_num(n):
    print(f'{n} is a perfect number')
else:
    print(f'{n} is NOT perfect number')
```

**20. Write a program to reverse a string word by word.**

```python
def reverse_words(sentence):
    
    words = sentence.split()
    rev_words = words[:: -1]
    return ' '.join(rev_words)
    return " ".join(sentence.split()[::-1])

original = 'The quick brown fox'
result = reverse_words(original)
print(result)
```
