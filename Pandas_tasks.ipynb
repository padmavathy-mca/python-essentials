```python
#import the libraries
import pandas as pd
import numpy as np
```

```python
#creating a dataset in a dictionary format (key = columns, values = rows)
data = {
    "Order_ID": [101, 102, 103, 104, 105, 106],
    "Product": ["Laptop", "Mobile", "Laptop", "Mouse", "Mobile", "Keyboard"],
    "Category": ["Electronics", "Electronics", "Electronics",
                 "Accessories", "Electronics", "Accessories"],
    "City": ["Chennai", "Chennai", "Bangalore",
             "Chennai", "Bangalore", "Chennai"],
    "Quantity": [2, 3, 1, 5, 2, 4],
    "Price": [50000, 20000, 50000, 500, 20000, 1000]
}

#inserting data into a dataframe 
df = pd.DataFrame(data)

#printing the dataframe
print(df)
```

```python
#Series - single column of data 
#creating a new series of columns
sales = pd.Series([10000, 20000, 15000, 30000],index=['Laptop','Mobile','Keyboard','Mouse'])
print('Sales of Products')
print(sales)
```

```python
#Access single column 
print(df["Product"])
```

```python
#Accessing multiple columns
print(df[["Product", "Quantity"]])
```

```python
#Sales = Quantity × Price
quantity = np.array(df['Quantity'])
price = np.array(df['Price'])
sales = quantity * price

print(sales)
```

```python
df["Sales"] = sales

print(df)
```

```python
#Access using labels
print(df.loc[0, "Product"])
```

```python
#Access a particular row
print(df.iloc[0])
```

```python
#Access specific rows
print(df.iloc[0:3])
```

```python
#Show me the product and quantity sold for every order
print(df[["Order_ID", "Product", "Quantity"]])
```

```python
#Find laptop sales
laptop_sales = df[df['Product'] == 'Laptop']
print(laptop_sales)
```

```python
#Find chennai sales
chennai_sales = df[df['City'] == 'Chennai']
print(chennai_sales)
```

```python
#Quatity greater than 2
high_quantity = df[df['Quantity'] > 2]
print(high_quantity)
```

```python
#Multiple conditions
#Find Electronics sales from Chennai
result = df[(df['Category'] == 'Electronics') & (df['City'] == 'Chennai')]
print(result)
```

```python
#Sort by quantity
sort_by_quantity = df.sort_values('Quantity')
print(sort_by_quantity)
```

```python
#Sort by quantity (descending)
sort_by_quantity = df.sort_values('Quantity',ascending=False)
print(sort_by_quantity)
```

```python
#which order contains the highest quantity
print(df.sort_values('Quantity',ascending=False)[["Order_ID", "Product", "Quantity"]])
```

```python
#Total quantity sold
total_quantity = df['Quantity'].sum()
print(total_quantity)
```

```python
#Average Selling price
avg_price = df['Price'].mean()
print(f'{avg_price:.2f}')
```

```python
#Maximum Quantity
print(df['Quantity'].max())
```

```python
#Minimum Quantity
print(df['Quantity'].min())
```

```python
#Median Quantity
print(df['Quantity'].median())
```

```python
#Number of Orders
print(df['Order_ID'].count())
```

```python
#Total sales by product
product_sales = df.groupby('Product')['Sales'].sum()
print(product_sales)
```

```python
#Total sales by City
city_sales = df.groupby('City')['Sales'].sum()
print(city_sales)
```

```python
#Quantity sold by Category
category_quantity = df.groupby('Category')['Quantity'].sum()
print(category_quantity)
```

```python
#Multiple Statistics
result = df.groupby('Category')['Sales'].agg(['sum','mean','max'])
print(result)
```
