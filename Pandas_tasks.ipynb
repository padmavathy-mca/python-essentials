```python
#Create missing values
df_missing = df.copy()

df_missing.loc[2,'City'] = np.nan
df_missing.loc[4,'Price'] = np.nan
df_missing.loc[5,'Category'] = np.nan

df_missing
```

```python
#Find the number of missing values in each column
df_missing.isnull().sum()
```

```python
#Display only the rows containing missing values
df_missing[df_missing.isnull().any(axis=1)]
```

```python
#Fill missing City with "Unknown"
df_missing['City'] = df_missing['City'].fillna('Unknown')

df_missing.loc[2,'City']
```

```python
#Fill missing Category with "Unknown"
df_missing['Category'] = df_missing['Category'].fillna('Unknown')

df_missing.loc[5,'Category']
```

```python
#Replace missing Price with the average price
df_missing['Price'] = df_missing['Price'].fillna(df_missing['Price'].mean())

df_missing.loc[4,'Price'].item()
```

```python
#Verify that no missing values remain
df_missing[df_missing.notnull().all(axis = 1)]
```

```python
#Create a Discount column with a 10% discount
df['Discount'] = df['Price'] * 0.10
df
```

```python
#Final_Amount = Total_Amount - Discount
df['Final_amount'] = df['Price'] - df['Discount']

df
```

```python
#Remove the Discount column
df.drop('Discount', axis=1 , inplace = True)

df
```

```python
#Renaming the columns
df.rename(columns={
    "Order_ID": "OrderID",
    "Price": "Unit_Price",
    "Category": "Product_Category"
}, inplace=True)

df
```
